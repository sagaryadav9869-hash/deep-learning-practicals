# Practical 2: NumPy, Pandas and Matplotlib

**Aim:** Practise data handling and visualisation on the Titanic dataset: filtering, missing values, group-by, merging, NumPy operations and Matplotlib charts.

```python
# -------------------------------
# IMPORT LIBRARIES
# -------------------------------
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
# -------------------------------
# LOAD DATASET
# -------------------------------
df = pd.read_csv("Titanic-Dataset .csv")
print("FIRST 5 ROWS:")
display(df.head(5))
print("LAST 5 ROWS:")
display(df.tail(5))
```

```python
# -------------------------------
# FILTERING DATA
# -------------------------------
df_filtered = df[(df['Age'] > 30) & (df['Survived'] == 1)]
print("Passengers Age > 30 and Survived:")
display(df_filtered)
# -------------------------------
```

```python
# HANDLING MISSING VALUES
# -------------------------------
print("Filling missing values with 0:")
df_fill = df.fillna(0)
display(df_fill)
print("Dropping rows with missing values:")
df_drop = df.dropna()
display(df_drop)
print("Interpolated missing values:")
df_interp = df.interpolate()
display(df_interp)
print("Missing Values Count:")
print(df.isnull().sum())
```

```python
# -------------------------------
# GROUP-BY OPERATIONS
# -------------------------------
print("\nMean Age by Passenger Class:")
print(df.groupby("Pclass")["Age"].mean())
print("\nCount of Sex by Survived:")
print(df.groupby("Survived")["Sex"].count())
print("\nFare (Mean, Max, Min) by Pclass:")
print(df.groupby("Pclass")["Fare"].agg(["mean", "max", "min"]))
```

```python
# -------------------------------
# MERGING DATAFRAMES
# -------------------------------
df1 = df[["PassengerId", "Survived", "Pclass"]]
df2 = df[["PassengerId", "Age", "Fare"]]
merged_inner = pd.merge(df1, df2, on="PassengerId", how="inner")
print("Inner Merge:")
display(merged_inner)
merged_left = pd.merge(df1, df2, on="PassengerId", how="left")
print("Left Merge:")
display(merged_left)
merged_right = pd.merge(df1, df2, on="PassengerId", how="right")
print("Right Merge:")
display(merged_right)
merged_outer = pd.merge(df1, df2, on="PassengerId", how="outer")
print("Outer Merge:")
display(merged_outer)
# -------------------------------
```

```python
# NUMPY ARRAY CREATION
# -------------------------------
arr1 = np.array([1, 2, 3, 4])
print("np.array:", arr1)
arr2 = np.arange(0, 10, 2)
print("np.arange:", arr2)
arr3 = np.linspace(0, 1, 5)
print("np.linspace:", arr3)
arr4 = np.random.randn(3, 3)
print("Random Matrix:\n", arr4)
```

```python
# Arithmetic
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])
print("Add:", a + b)
print("Subtract:", a - b)
print("Multiply:", a * b)
print("Divide:", a / b)
print("Square:", a ** 2)
```

```python
# Matrix operations
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
print("Dot:\n", np.dot(A, B))
print("Matmul:\n", np.matmul(A, B))
print("@ Operator:\n", A @ B)
```

```python
# Reshaping
arr = np.arange(1, 10)
print("Reshape:\n", arr.reshape(3, 3))
print("Ravel:", arr.ravel())
print("Flatten:", arr.flatten())
```

```python
# Statistics
nums = np.array([10, 20, 30, 40, 50])
print("Mean:", np.mean(nums))
print("Median:", np.median(nums))
print("Variance:", np.var(nums))
print("Std Dev:", np.std(nums))
```

```python
# Random distributions
print("Uniform:", np.random.rand(5))
print("Normal:", np.random.randn(5))
print("Random Ints:", np.random.randint(10, 50, 5))
# -------------------------------
```

```python
# MATPLOTLIB VISUALIZATION
# -------------------------------
# Line Chart
plt.plot(df["PassengerId"], df["Age"])
plt.title("Passenger Age Line Chart")
plt.xlabel("Passenger ID")
plt.ylabel("Age")
plt.grid(True)
plt.show()
```

```python
# Bar Chart
surv_counts = df["Survived"].value_counts()
plt.bar(surv_counts.index, surv_counts.values)
plt.title("Survived vs Not Survived")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Count")
plt.show()
```

```python
# Pie Chart
pclass_counts = df["Pclass"].value_counts()
plt.pie(pclass_counts, labels=pclass_counts.index, autopct="%1.1f%%")
plt.title("Passenger Class Distribution")
plt.show()
```

```python
# Scatter Plot
plt.scatter(df["Age"], df["Fare"])
plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()
```

```python
# Histogram
plt.hist(df["Age"].dropna(), bins=30)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```
