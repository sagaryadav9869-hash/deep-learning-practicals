# Practical 1: McCulloch-Pitts Neuron

**Aim:** Implement the McCulloch-Pitts neuron model and use it to build the AND, OR, NOT, NAND and NOR logic gates.

```python
import numpy as np

class McCullochPittsNeuron:
    def __init__(self, weights, threshold):
        self.weights = np.array(weights)
        self.threshold = threshold

    def activate(self, inputs):
        # Calculate the weighted sum of inputs
        weighted_sum = np.dot(inputs, self.weights)
        # Apply the threshold activation function
        return 1 if weighted_sum >= self.threshold else 0

# AND Gate
# Output is 1 only if both inputs are 1
# Weights: [1, 1], Threshold: 2
and_gate = McCullochPittsNeuron(weights=[1, 1], threshold=2)
print("AND Gate:")
print(f"0 AND 0: {and_gate.activate([0, 0])}")
print(f"0 AND 1: {and_gate.activate([0, 1])}")
print(f"1 AND 0: {and_gate.activate([1, 0])}")
print(f"1 AND 1: {and_gate.activate([1, 1])}")
print("-" * 20)

# OR Gate
# Output is 1 if at least one input is 1
# Weights: [1, 1], Threshold: 1
or_gate = McCullochPittsNeuron(weights=[1, 1], threshold=1)
print("OR Gate:")
print(f"0 OR 0: {or_gate.activate([0, 0])}")
print(f"0 OR 1: {or_gate.activate([0, 1])}")
print(f"1 OR 0: {or_gate.activate([1, 0])}")
print(f"1 OR 1: {or_gate.activate([1, 1])}")
print("-" * 20)

# NOT Gate
# Output is the opposite of the input
# Weights: [-1], Threshold: 0
not_gate = McCullochPittsNeuron(weights=[-1], threshold=0)
print("NOT Gate:")
print(f"NOT 0: {not_gate.activate([0])}")
print(f"NOT 1: {not_gate.activate([1])}")
print("-" * 20)

# NAND Gate
# Output is 0 only if both inputs are 1
# Weights: [-1, -1], Threshold: -1
nand_gate = McCullochPittsNeuron(weights=[-1, -1], threshold=-1)
print("NAND Gate:")
print(f"0 NAND 0: {nand_gate.activate([0, 0])}")
print(f"0 NAND 1: {nand_gate.activate([0, 1])}")
print(f"1 NAND 0: {nand_gate.activate([1, 0])}")
print(f"1 NAND 1: {nand_gate.activate([1, 1])}")
print("-" * 20)

# NOR Gate
# Output is 1 only if both inputs are 0
# Weights: [-1, -1], Threshold: 0
nor_gate = McCullochPittsNeuron(weights=[-1, -1], threshold=0)
print("NOR Gate:")
print(f"0 NOR 0: {nor_gate.activate([0, 0])}")
print(f"0 NOR 1: {nor_gate.activate([0, 1])}")
print(f"1 NOR 0: {nor_gate.activate([1, 0])}")
print(f"1 NOR 1: {nor_gate.activate([1, 1])}")
print("-" * 20)
```
